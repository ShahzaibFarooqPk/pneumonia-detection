```python
import os

model_path = "custom_pre_trained_model_10.keras"

print("Model exists:", os.path.exists(model_path))
```

```python
import tensorflow as tf

model = tf.keras.models.load_model(model_path)

print("Model loaded successfully!")

print("Input shape:", model.input_shape)
print("Output shape:", model.output_shape)
```

```python
from PIL import Image
from IPython.display import display

img = Image.open("normal_img_pneomonia.jpg")

display(img)
```

```python
# Convert image into RGB
img = img.convert("RGB")

# Resize image
img = img.resize((100, 100))

# Convert image into NumPy array
new_array = tf.keras.utils.img_to_array(img)

# Add batch dimension
new_array = np.expand_dims(new_array, axis=0)

# Normalize pixel values
new_array = new_array / 255.0

print("Image shape:", new_array.shape)
```

```python
prediction = model.predict(new_array)

print("Prediction:", prediction)
```

```python
# Check the final layer activation
print("Final Activation:",
      model.layers[-1].activation.__name__)

# Check model input shape
print("Input Shape:", model.input_shape)

# Check model output shape
print("Output Shape:", model.output_shape)

# Check prediction
print("Prediction:", prediction[0][0])
```
